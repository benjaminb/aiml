```python
from langchain.messages import SystemMessage, HumanMessage
from pydantic import BaseModel, Field


class ListOfStringsModel(BaseModel):
    items: list[str] = Field(..., description="A list of strings")
```

```python
from structured_llm import StructuredLLM


sllm = StructuredLLM(
    model="gpt-5-nano",
    response_model=ListOfStringsModel
)
```

```python
prompt = "Write a list of three fruits"
response = sllm.invoke(prompt)

print(f"Parsed response: {response['parsed']}")
print(f"Raw response: {response['raw']}")
```

```python


model_name = "ollama:ministral-3:3b"
chat_model = init_chat_model(
    model=model_name,
)
```

```python
print(chat_model.model)
```

```python
response = chat_model.invoke([SystemMessage(content="Write a list of three fruits")])
```

```python
print(response)
```

```python
structured_chat_model = chat_model.with_structured_output(
    ListOfStringsModel, include_raw=True)


system_message = SystemMessage(
    content="You are a helpful assistant that returns a list of strings. Write a list of three fruits.")

messages = [system_message]

# Anthropic models require a human message
if model_name.startswith("claude"):
    messages.append(HumanMessage(content="do it"))

response = structured_chat_model.invoke(messages)
```

```python
print(response['parsed'].items)
```
